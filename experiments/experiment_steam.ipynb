```python
# Load processed data (run this if data preprocessing was done previously)
final_df = pd.read_csv('data/steam/user_title_timestamp_with_names.csv')
print(f"📊 Loaded processed Steam data: {len(final_df):,} records")
print(f"📈 Unique users: {final_df['UserID'].nunique():,}")
print(f"🎮 Unique games: {final_df['Title'].nunique():,}")
print("\n📊 Sample of processed data:")
print(final_df.head())
```

```python
# Initialize the LLM Position Bias Analyzer for Steam games
print("🚀 Initializing LLM Position Bias Analyzer for Steam Games...")

analyzer = LLMPositionBiasAnalyzer(
    data=final_df,
    data_name="steam",
    model="gpt-3.5-turbo",
    backend="openai",
    list_size=20,
    api_tier="tier_2"
)
    
print("✅ Analyzer initialized successfully!")
```

```python
# bias_analysis =  analyzer.compute_bias_analysis(5,None,True,3,20)
# print(bias_analysis)
```

```python
# Pre-calculated bias scores for Steam dataset (to be updated after running bias analysis)
prebias_gpt4o_steam = {
    'avg_primacy': 0.41600000000000004,
    'avg_recency': 0.26799999999999996,
    'avg_middle': 1.316
}

prebias_gpt3_steam = {
    'avg_primacy': 0.5159999999999999,
    'avg_recency': 0.496,
    'avg_middle': 0.9880000000000001
}
```

```python
# Main debiasing experiment with list size 20
print("\n🔧 COMPLETE DEBIASING EXPERIMENT - STEAM GAMES")
print("=" * 55)

num_candidates = 25    # Number of candidates per evaluation
num_trials = 5        # Number of randomization trials per user


results = analyzer.evaluate_our_method_batched(
    num_candidates=num_candidates,
    num_trials=num_trials,
    aggregation_method="mean",
    use_parallel=True,
    precalculated_bias=prebias_gpt3_steam,
    checkpoint_file="evaluation_checkpoint_steam_gpt3_trails25.json"
)
```

```python
# Analysis and visualization of results
print("📊 Analyzing Steam debiasing results...")

# Load results from checkpoint if available
checkpoint_file = "evaluation_checkpoint_steam.json"
if os.path.exists(checkpoint_file):
    with open(checkpoint_file, 'r') as f:
        results = json.load(f)
    
    print("✅ Loaded results from checkpoint")
    print(f"📈 Total users evaluated: {len(results.get('user_results', {}))}")
    print(f"🎮 Dataset: Steam Games")
    print(f"🤖 Model: GPT-3.5-turbo")
    
    # Display key metrics
    if 'aggregated_results' in results:
        agg_results = results['aggregated_results']
        print("\n📊 Key Metrics:")
        print(f"🎯 Average NDCG: {agg_results.get('avg_ndcg', 'N/A'):.4f}")
        print(f"🎯 Average Precision: {agg_results.get('avg_precision', 'N/A'):.4f}")
        print(f"🎯 Average Recall: {agg_results.get('avg_recall', 'N/A'):.4f}")
        print(f"🎯 Average MRR: {agg_results.get('avg_mrr', 'N/A'):.4f}")
else:
    print("⚠️  No checkpoint file found. Run the evaluation first.")
```

```python
# Save final results summary
print("💾 Saving experiment summary...")

experiment_summary = {
    "dataset": "steam",
    "model": "gpt-3.5-turbo",
    "experiment_date": datetime.now().isoformat(),
    "num_candidates": 20,
    "num_trials": 20,
    "batch_size": 20,
    "total_users": len(final_df['UserID'].unique()) if 'final_df' in locals() else 0,
    "total_games": len(final_df['Title'].unique()) if 'final_df' in locals() else 0,
    "checkpoint_file": "evaluation_checkpoint_steam.json"
}

with open('steam_experiment_summary.json', 'w') as f:
    json.dump(experiment_summary, f, indent=2)

print("✅ Experiment summary saved to steam_experiment_summary.json")
print("\n🎮 Steam debiasing experiment completed!")
```
